```python
fruits = ["apple", "mango", "lichi", 121, 1.2451, True]
fruits
```

```python
#HOW TO ACCESS THE LIST
#INDEXING in python starts from 0
fruits[0] #index 0 is the first element , the first element is apple
fruits[-1]
```

```python
#To add elements in a list
fruits.append("banana")
fruits.append(5464)
fruits
```

```python
#removing items from list
fruits.remove(5464)
fruits
```

```python
#iterating through lists
for i in fruits:
  print(i)
```

```python
#length function
len(fruits)
```

```python
#exsistence of a particular item in a list
print("mango" in fruits)
print("cocnut" in fruits)
```
