```python
#Set in python is the unordered collections of unique items
```

```python
#Creation of set
my_set = {1,2,3,4,5,6,6,6,6,}
my_set #set will remove all duplicate items from the set
```

```python
#Adding elements to the set
my_set.add(9)
my_set.add(4)
print(my_set)
```

```python
#deleting certain entries from the set
my_set.discard(1) #Deleting 1 from the set (using discard to avoid KeyError if 1 is not present)
print(my_set)
my_set.discard(6) #Corrected: using discard() instead of non-existent delete() method
print(my_set)
```

```python
my_set.remove(5) #Deleting 5 from the set
print(my_set)
```

```python
#Operations in set
set1 = {1,2,3,4,5,6}
set2 = {6,7,8,9,10,11,12}
print(set1.union(set2)) #UNION OF SETS
print(set1.intersection(set2)) #INTERSECTION OF SETS
print(set1.difference(set2)) #DIFFERENCE OF SETS
print(set2.difference(set1))
```
